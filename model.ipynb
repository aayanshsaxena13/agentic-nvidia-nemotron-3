# **Agentic Nvidia Nemotron 3 Super**

```python
import os
from openrouter import OpenRouter
import dotenv
import json

dotenv.load_dotenv()

apiKey = os.getenv("API_KEY")
client = OpenRouter(api_key=apiKey)

rules = ""
skill = open("Agents.md", "r")
rules = skill.read()
skill.close()

while True:
    prompt = f"{input("You:")}"

    response = client.chat.send(
        model="nvidia/nemotron-3-super-120b-a12b:free",
        messages=[{ "role": "user", "content": f"{rules} {prompt}" }],
    )

    response = response.choices[0].message.content
    response = json.loads(response)

    for x in response:
        dirname = os.path.dirname(x['filePath'])
        if dirname:
            os.makedirs(dirname, exist_ok=True)

        with open(x['filePath'], "w") as f:
            f.write(x['code'])
    
    print('Nvidia Nemotron 3 Super: Project has successfully been completed.')
```
